```python
from google_play_scraper import app

result = app(
    'mam.reader.ipusnas',
    lang='en',
    country='us'
)

print(f"App: {result['title']}")
print(f"Rating: {result['score']}")
print(f"Downloads: {result['installs']}")
print(f"Price: {result['free']}")
```

```python
from google_play_scraper import reviews, Sort

result, continuation_token = reviews(
    'mam.reader.ipusnas',
    lang='en',
    country='us',
    sort=Sort.NEWEST,
    count=4000
)
for review in result:
  print(f"{review['userName']}: {review['score']} stars")
  print(f"Review: {review['content']}")
  print(f"Date: {review['at']}")
```

```python
import pandas as pd
df=pd.DataFrame(result)
df.to_csv('ipusnasdata_raw.csv', index=False)
```

```python
df['at'] = pd.to_datetime(df['at'])
start = pd.to_datetime("2021-01-01")
end   = pd.to_datetime("2026-05-16")
mask = (df['at'] >= start) & (df['at'] <= end)
filtered = df[mask]

filtered[['userName','score','content','at']]
```

```python
filtered.to_csv("rawipusnasdata.csv", index=False)
```
